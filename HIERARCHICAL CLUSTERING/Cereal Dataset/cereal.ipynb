```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
```

```python
data = pd.read_csv("cereal.csv")
```

```python
X = data[['calories','protein','fat','sodium','fiber','sugars']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
linked = linkage(X_scaled, method='single')
```

```python
plt.figure(figsize=(10,6))
dendrogram(linked)
plt.title("Dendrogram - Cereal Dataset")
plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=3, linkage='single')
labels = model.fit_predict(X_scaled)
```

```python
data['Cluster'] = labels
print(data[['name','Cluster']].head())
```
